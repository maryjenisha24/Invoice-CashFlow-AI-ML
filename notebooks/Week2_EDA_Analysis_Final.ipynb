## 1. Dataset Loading and Overview

```python
import pandas as pd
from pathlib import Path

project_root = next(
    (p for p in [Path.cwd(), *Path.cwd().parents]
     if (p / 'data' / 'cleaned' / 'cleaned_dataset.csv').exists()),
    None
)

if project_root is None:
    raise FileNotFoundError('Project folder containing cleaned dataset not found.')

data_path = project_root / 'data' / 'cleaned' / 'cleaned_dataset.csv'
df = pd.read_csv(data_path)

print('Dataset Shape:', df.shape)
print('\nDataset Information:')
df.info()
display(df.head())
print('\nMissing Values:')
display(df.isnull().sum())
```
